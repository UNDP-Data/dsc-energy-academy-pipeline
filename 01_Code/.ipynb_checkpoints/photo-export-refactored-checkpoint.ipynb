```python
# ✅ ENV SETUP
import os
from dotenv import load_dotenv

load_dotenv(".env")

FIGMA_TOKEN = os.getenv("FIGMA_TOKEN")
FILE_ID_1 = os.getenv("FIGMA_DOCUMENT_ID1")
FILE_ID_2 = os.getenv("FIGMA_DOCUMENT_ID2")
FILE_ID_3 = os.getenv("FIGMA_DOCUMENT_ID3")
SAS_URL = os.getenv("SAS_URL")

structure_path = "../03_Outputs/SEA_Modules/en/module_structure.json"
figma_path_1 = "../02_Inputs/figma_jsons/figma_document1.json"
figma_path_2 = "../02_Inputs/figma_jsons/figma_document2.json"
figma_path_3 = "../02_Inputs/figma_jsons/figma_document3.json"

cache_path = "../03_Outputs/image_refs_cache.json"

pages_1=["Module 1", "Module 2", "Module 3"]
pages_2=["Module 4", "Module 5", "Module 6"]
pages_3=["Module 7", "Module 8", "Module 9"]

required = ["FIGMA_TOKEN", "FILE_ID_1", "FILE_ID_2", "FILE_ID_3","SAS_URL"]
missing = [v for v in required if not globals()[v]]
if missing:
    raise EnvironmentError(f"❌ Missing env vars: {', '.join(missing)}")

print("✅ Environment loaded.")

# ✅ UTILS
import io
import time
from PIL import Image
from contextlib import contextmanager
```

```python


def format_blob_name(folder: str, name: str) -> str:
    return f"{folder}/{name.replace(':', '_').replace(' ', '_')}.webp"

def format_path(name: str) -> str:
    return name.replace(" ", "_").strip()

@contextmanager
def timed(label):
    t0 = time.time()
    yield
    print(f"⏱️ {label} took {time.time() - t0:.2f}s")

def convert_image_bytes_to_webp(image_bytes: bytes, max_height: int = 1080, quality: int = 95) -> bytes:
    with Image.open(io.BytesIO(image_bytes)) as img:
        img = img.convert("RGB")

        # Calculate scale factor if height is above max_height
        if img.height > max_height:
            scale = max_height / img.height
            new_width = int(img.width * scale)
            new_height = max_height
            img = img.resize((new_width, new_height), Image.Resampling.LANCZOS)

        buffer = io.BytesIO()
        img.save(buffer, format="WEBP", quality=quality)
        return buffer.getvalue()


# ✅ AZURE STORAGE
import ssl
from azure.storage.blob.aio import ContainerClient
from azure.storage.blob import ContentSettings
from azure.core.pipeline.transport import AioHttpTransport

USE_INSECURE_SSL = False

def get_container_client():
    if USE_INSECURE_SSL:
        ssl_context = ssl.create_default_context()
        ssl_context.check_hostname = False
        ssl_context.verify_mode = ssl.CERT_NONE
        transport = AioHttpTransport(ssl_context=ssl_context)
        return ContainerClient.from_container_url(SAS_URL, transport=transport)
    return ContainerClient.from_container_url(SAS_URL)

async def get_existing_blob_names(prefix: str = "") -> set[str]:
    blob_names = set()
    async with get_container_client() as client:
        async for blob in client.list_blobs(name_starts_with=prefix):
            blob_names.add(blob.name)
    print(f"📦 {len(blob_names)} blobs found with prefix '{prefix}'")
    return blob_names

async def upload_image_blob(blob_name: str, data: bytes) -> str:
    async with get_container_client() as client:
        blob = client.get_blob_client(blob_name)
        await blob.upload_blob(
            data=data,
            overwrite=True,
            content_settings=ContentSettings(content_type="image/webp"),
        )
        return blob.url.split("?")[0]

def assign_parents(node: dict, parent: dict | None = None):
    node["_parent"] = parent
    for child in node.get("children", []) or []:
        assign_parents(child, node)


# ✅ FETCH IMAGE URLS
import requests
import json

REFS_PATH = "figma_image_refs.json"

def calculate_optimal_scale(canvas_height: float, target_height: int = 1080, max_scale: float = 4.0) -> float:
    """
    Calculate the optimal scale factor for Figma image export.

    Parameters:
    - canvas_height (float): The height of the image on the canvas.
    - target_height (int): The desired minimum height in pixels after scaling.
    - max_scale (float): The maximum scale Figma API allows.

    Returns:
    - float: A scale factor (rounded to 2 decimals) to use in the image export URL.
    """
    if not canvas_height or canvas_height <= 0:
        return 2.0  # sensible default if height is missing

    scale = target_height / canvas_height
    adjusted_scale=round(min(max_scale, max(1.0, scale)), 2)
    return adjusted_scale


def fetch_image_url_single(file_id: str, node_id: str, canvas_height: float, format="png") -> str | None:
    try:
        scale = calculate_optimal_scale(canvas_height)
        res = requests.get(
            f"https://api.figma.com/v1/images/{file_id}",
            headers={"X-Figma-Token": FIGMA_TOKEN},
            params={"ids": node_id, "format": format, "scale": scale},
            timeout=30,
        )
        res.raise_for_status()
        return res.json().get("images", {}).get(node_id)
    except Exception as e:
        print(f"❌ Error fetching URL for {node_id}: {e}")
        return None



def load_image_refs() -> dict[str, str]:
    if os.path.exists(REFS_PATH):
        with open(REFS_PATH, "r") as f:
            return json.load(f)
    return {}

def save_image_refs(refs: dict[str, str]):
    with open(REFS_PATH, "w") as f:
        json.dump(refs, f, indent=2)

# ✅ FETCH IMAGE CONTENT
import aiohttp

async def fetch_image_content(url: str) -> bytes | None:
    try:
        async with aiohttp.ClientSession() as session:
            async with session.get(url, timeout=30) as resp:
                if resp.status == 200:
                    return await resp.read()
                print(f"⚠️ Failed to fetch {url} (status {resp.status})")
    except Exception as e:
        print(f"❌ Error fetching image: {e}")
    return None

# ✅ SCAN FIGMA PAGE

from math import ceil

def extract_generic_image(node: dict, page_name: str) -> dict | None:
    if node.get("type") != "RECTANGLE":
        return None
    for fill in node.get("fills", []):
        if fill.get("type") == "IMAGE" and "imageRef" in fill:
            canvas_height = node.get("absoluteBoundingBox", {}).get("height", 0)
            return {
                "node_id": node["id"],
                "blob_name": format_blob_name(f"Modules/{format_path(page_name)}", node["id"]),
                "image_ref": fill["imageRef"],
                "canvas_height": node.get("absoluteBoundingBox", {}).get("height", 0),
                "type": "generic"
            }
    return None


def scan_figma_page(page):
    scanned = {}
    assign_parents(page)
    type_counts = {"generic": 0}

    def walk(node):
        if node.get("type") == "RECTANGLE":
            result = extract_generic_image(node, page["name"])
            if result:
                scanned[result["node_id"]] = result
                type_counts["generic"] += 1
        for child in node.get("children", []) or []:
            walk(child)

    walk(page)
    total = sum(type_counts.values())
    print(f"🔍 {page['name']}: {total} image rectangles found")
    return scanned


# ✅ UPLOAD

# ✅ UPLOAD (UPDATED)
async def upload_figma_images(
    file_id,
    id_to_blobname,
    image_refs,
    id_to_height,          # ✅ NEW: node_id -> canvas_height
    existing_blobs,
    cache_path,
    overwrite=False,
):
    previous_refs = {}
    if os.path.exists(cache_path):
        try:
            with open(cache_path, "r") as f:
                previous_refs = json.load(f)
        except Exception as e:
            print(f"❌ Failed to load cache: {e}")

    to_upload = {
        node_id: blob_name
        for node_id, blob_name in id_to_blobname.items()
        if overwrite
        or blob_name not in existing_blobs
        or image_refs.get(node_id) != previous_refs.get(node_id)
    }

    print(f"🧮 {len(id_to_blobname)} total | {len(to_upload)} to upload")

    uploaded = 0
    skipped_no_height = 0
    skipped_no_url = 0
    skipped_no_bytes = 0

    for i, (node_id, blob_name) in enumerate(to_upload.items(), 1):
        canvas_height = id_to_height.get(node_id, 0)
        if not canvas_height:
            skipped_no_height += 1
            # not fatal, but usually indicates inconsistent mappings
            # (still try with height=0, which yields default scale in calculate_optimal_scale)
            # If you prefer hard-skip, uncomment next two lines:
            # print(f"⚠️ Missing canvas_height for {node_id}; skipping")
            # continue

        url = fetch_image_url_single(file_id, node_id, canvas_height)
        if not url:
            skipped_no_url += 1
            continue

        image_bytes = await fetch_image_content(url)
        if not image_bytes:
            skipped_no_bytes += 1
            continue

        webp = convert_image_bytes_to_webp(image_bytes)
        print(f"✅ [{i}] Uploading {blob_name}")
        await upload_image_blob(blob_name, webp)

        # update cache (only if we have a ref for this node)
        if node_id in image_refs:
            previous_refs[node_id] = image_refs[node_id]
            try:
                with open(cache_path, "w") as f:
                    json.dump(previous_refs, f, indent=2)
            except Exception as e:
                print(f"❌ Failed to update cache after {blob_name}: {e}")

        uploaded += 1

    print(
        f"📊 Done | Uploaded: {uploaded} | "
        f"Skipped(no_url): {skipped_no_url} | Skipped(no_bytes): {skipped_no_bytes} | MissingHeight: {skipped_no_height}"
    )

with open(structure_path) as f:
    structure_json = json.load(f)
with open(figma_path_1) as f:
    figma_data_1 = json.load(f)
with open(figma_path_2) as f:
    figma_data_2 = json.load(f)
with open(figma_path_3) as f:
    figma_data_3 = json.load(f)

file_page_sets = [
    (FILE_ID_1, figma_data_1["children"], pages_1),
    (FILE_ID_2, figma_data_2["children"], pages_2),
    (FILE_ID_3, figma_data_3["children"], pages_3)
]


print("📦 Fetching existing Azure blobs...")
existing_blobs = await get_existing_blob_names()

print("🔄 Loading cached image references...")
updated_refs = load_image_refs()
```

```python

# ✅ 8. MAIN PIPELINE RUNNER (UPDATED CALL)
for file_id, figma_pages, page_names in file_page_sets:
    for page_name in page_names:
        page = next((p for p in figma_pages if p["name"] == page_name), None)
        if not page:
            print(f"⚠️ Page not found: {page_name}")
            continue

        print(f"\n📄 Processing page: {page_name}")
        scanned = scan_figma_page(page)
        if not scanned:
            continue

        id_to_blobname = {
            node_id: info["blob_name"].replace(" ", "_")
            for node_id, info in scanned.items()
        }

        fetched_refs = {
            node_id: info["image_ref"]
            for node_id, info in scanned.items()
            if "image_ref" in info
        }

        id_to_height = {
            node_id: float(info.get("canvas_height", 0) or 0)
            for node_id, info in scanned.items()
        }

        await upload_figma_images(
            file_id=file_id,
            id_to_blobname=id_to_blobname,
            image_refs=fetched_refs,
            id_to_height=id_to_height,       # ✅ NEW
            existing_blobs=existing_blobs,
            cache_path=cache_path,
            overwrite=True,
        )

print("✅ Upload pipeline complete.")
```

```python
# ✅ INFOGRAPHIC EXPORT CELL (updated for new upload_figma_images signature)
# - Passes id_to_height (no reliance on global `scanned`)
# - Keeps clean logging via redirect_stdout
# - overwrite=True
# - No intermediate prints except per-item + final summary

import re
import io
from contextlib import redirect_stdout

def find_figma_page(figma_pages: list[dict], page_name: str) -> dict | None:
    for p in figma_pages:
        if p.get("name") == page_name and p.get("type") in ("CANVAS", "PAGE", "FRAME"):
            return p
    return None

def find_first_node_by_name(node: dict, target_name: str) -> dict | None:
    if node.get("name") == target_name:
        return node
    for ch in node.get("children", []) or []:
        hit = find_first_node_by_name(ch, target_name)
        if hit:
            return hit
    return None

def walk_collect(node: dict, out: list):
    out.append(node)
    for ch in node.get("children", []) or []:
        walk_collect(ch, out)

def get_canvas_height(node: dict, fallback: float = 800.0) -> float:
    bb = node.get("absoluteBoundingBox") or {}
    h = bb.get("height")
    try:
        h = float(h)
    except Exception:
        h = None
    return h if h and h > 0 else float(fallback)

total_uploaded = 0
total_skipped  = 0
per_item_logs  = []

for file_id, figma_pages, page_names in file_page_sets:
    for page_name in page_names:
        page = find_figma_page(figma_pages, page_name)
        if not page:
            continue

        nodes = []
        walk_collect(page, nodes)

        infographic_frames = [
            n for n in nodes
            if n.get("type") == "FRAME" and n.get("name") == "infographic"
        ]
        if not infographic_frames:
            continue

        for inf in infographic_frames:
            design = find_first_node_by_name(inf, "infographic-design")
            if not design:
                total_skipped += 1
                per_item_logs.append(f"⏭️ Skipped Modules/{format_path(page_name)}/(missing infographic-design)")
                continue

            node_id = design["id"]
            blob_name = f"Modules/{format_path(page_name)}/{node_id.replace(':','_')}.webp"

            id_to_blob   = {node_id: blob_name}
            image_refs   = {node_id: None}  # FRAME; caching not applicable
            id_to_height = {node_id: get_canvas_height(design, fallback=800.0)}  # ✅ NEW

            buf = io.StringIO()
            with redirect_stdout(buf):
                await upload_figma_images(
                    file_id=file_id,
                    id_to_blobname=id_to_blob,
                    image_refs=image_refs,
                    id_to_height=id_to_height,        # ✅ NEW
                    existing_blobs=existing_blobs,
                    cache_path=cache_path,
                    overwrite=True,
                )

            raw = buf.getvalue()

            # Parse our UPDATED final line:
            # "📊 Done | Uploaded: X | Skipped(no_url): A | Skipped(no_bytes): B | MissingHeight: C"
            m = re.findall(
                r"Done\s*\|\s*Uploaded:\s*(\d+)\s*\|\s*Skipped\(no_url\):\s*(\d+)\s*\|\s*Skipped\(no_bytes\):\s*(\d+)\s*\|\s*MissingHeight:\s*(\d+)",
                raw
            )
            if m:
                up, no_url, no_bytes, miss_h = map(int, m[-1])
                sk = no_url + no_bytes  # treat these as skipped for summary
            else:
                # Fallback heuristics (if logs change)
                up = 1 if "Uploading" in raw else 0
                sk = 1 - up

            total_uploaded += up
            total_skipped  += sk

            if up > 0:
                per_item_logs.append(f"✅ Uploaded {blob_name}")
            else:
                per_item_logs.append(f"⏭️ Skipped {blob_name}")

# Output: only per-item + final summary (no intermediate logs)
for line in per_item_logs:
    print(line)

print(f"🧾 Infographics: uploaded {total_uploaded}, skipped {total_skipped}")
print("✅ Infographic export complete.")
```

```python
# ✅ STANDALONE: EXCEL PHOTO INDEX (2 FIGMA DOCS) — TEST MODE (default 5 images)
# Run AFTER your ENV SETUP cell (the one that defines FILE_ID_1/2, figma_path_1/2, pages_1/2, and loads FIGMA_TOKEN).
# This cell:
#  - Loads cached Figma JSONs (figma_path_1, figma_path_2)
#  - Walks selected pages across both docs
#  - Finds RECTANGLE layers with IMAGE fills
#  - Derives module/chapter/lesson from PAGE + SECTION ancestry
#  - Fetches image bytes via your existing async fetchers (fetch_image_url_single + fetch_image_content)
#  - Writes an Excel with embedded thumbnails in Column A + metadata (including photo layer name)

import os, re, json
from io import BytesIO
from PIL import Image
from openpyxl import Workbook
from openpyxl.drawing.image import Image as XLImage

# =========================
# 🔧 SWITCHES YOU CHANGE
# =========================
TEST_MODE = True                 # True = quick test, False = full run
TEST_LIMIT = None                   # how many images to process in test mode
EXCEL_OUT_PATH = "../03_Outputs/photo_index_thumbs.xlsx"

THUMB_MAX_PX   = (220, 220)      # thumbnails are column A; keep them modest
THUMB_QUALITY  = 70
FIGMA_SCALE    = 2               # note: in your notebook, fetch_image_url_single uses (file_id, node_id, canvas_height)
# =========================


# ---------- Load cached Figma JSONs ----------
with open(figma_path_1, "r") as f:
    figma_data_1 = json.load(f)
with open(figma_path_2, "r") as f:
    figma_data_2 = json.load(f)

file_page_sets = [
    (FILE_ID_1, figma_data_1["children"], pages_1),
    (FILE_ID_2, figma_data_2["children"], pages_2),
]


# ---------- Utilities ----------
def px_to_excel_row_height(px: int) -> float:
    return px * 0.75  # px -> points (approx)

def px_to_excel_col_width(px: int) -> float:
    return px / 7.0   # rough but fine

def make_jpeg_thumbnail_bytes(image_bytes: bytes, max_px=(220,220), quality=70):
    with Image.open(BytesIO(image_bytes)) as im:
        im = im.convert("RGB")
        im.thumbnail(max_px)
        bio = BytesIO()
        im.save(bio, format="JPEG", quality=quality, optimize=True, progressive=True)
        bio.seek(0)
        return bio, im.size


# ---------- Hierarchy helpers ----------
def assign_parents(node: dict, parent=None):
    node["_parent"] = parent
    for c in node.get("children", []) or []:
        assign_parents(c, node)

def walk(node):
    yield node
    for ch in node.get("children", []) or []:
        yield from walk(ch)

def find_nearest_ancestor(node, types):
    cur = node.get("_parent")
    while cur:
        if cur.get("type") in types:
            return cur
        cur = cur.get("_parent")
    return None

def find_next_ancestor_section(start_node):
    cur = start_node.get("_parent") if start_node else None
    while cur:
        if cur.get("type") == "SECTION":
            return cur
        cur = cur.get("_parent")
    return None

def parse_module_number(name):
    if not name: return None
    m = re.search(r"\bModule\s*0*(\d+)\b", name, re.IGNORECASE)
    if m: return m.group(1)
    m = re.search(r"\bM\s*0*(\d+)\b", name, re.IGNORECASE)
    if m: return m.group(1)
    return None

def parse_chapter_number(name):
    if not name: return None
    m = re.search(r"\bChapter\s*0*(\d+)\b", name, re.IGNORECASE)
    if m: return m.group(1)
    m = re.search(r"\bC\s*0*(\d+)\b", name, re.IGNORECASE)
    if m: return m.group(1)
    return None

def parse_lesson_number(name):
    if not name: return None
    m = re.search(r"\bLesson\s*0*(\d+)\b", name, re.IGNORECASE)
    if m: return m.group(1)
    m = re.search(r"\bL\s*0*(\d+)\b", name, re.IGNORECASE)
    if m: return m.group(1)
    return None

def find_lesson_section(node):
    # nearest section might be a subsection; walk up until it looks like a lesson section
    sec = find_nearest_ancestor(node, ("SECTION",))
    while sec and not parse_lesson_number(sec.get("name","")):
        sec = find_next_ancestor_section(sec)
    return sec

def find_chapter_section_from_lesson(lesson_section):
    sec = find_next_ancestor_section(lesson_section) if lesson_section else None
    while sec and not parse_chapter_number(sec.get("name","")):
        sec = find_next_ancestor_section(sec)
    return sec

def is_image_rectangle(node):
    if node.get("type") != "RECTANGLE":
        return None
    for fill in node.get("fills") or []:
        if fill.get("type") == "IMAGE" and fill.get("imageRef"):
            return fill["imageRef"]
    return None

def find_page(figma_pages, name):
    for p in figma_pages:
        if p.get("name") == name:
            return p
    return None


# ---------- Workbook ----------
wb = Workbook()
ws = wb.active
ws.title = "PhotoIndex"

# Header row: THUMBNAIL IS COLUMN 1 (A)
ws.append([
    "thumbnail",            # A (embedded image)
    "file_id",              # B
    "module_number",        # C
    "chapter_number",       # D
    "lesson_number",        # E
    "page_name",            # F
    "chapter_section_name", # G
    "lesson_section_name",  # H
    "frame_name",           # I
    "photo_layer_name",     # J  ✅ NEW
    "node_id",              # K
    "image_ref",            # L
    "figma_image_url",      # M
])

# widths
ws.column_dimensions["A"].width = px_to_excel_col_width(THUMB_MAX_PX[0] + 14)  # thumbnail col
ws.column_dimensions["B"].width = 18
ws.column_dimensions["C"].width = 14
ws.column_dimensions["D"].width = 14
ws.column_dimensions["E"].width = 14
ws.column_dimensions["F"].width = 22
ws.column_dimensions["G"].width = 48
ws.column_dimensions["H"].width = 48
ws.column_dimensions["I"].width = 30
ws.column_dimensions["J"].width = 36
ws.column_dimensions["K"].width = 22
ws.column_dimensions["L"].width = 18
ws.column_dimensions["M"].width = 60

row_ptr = 2


# ---------- Runner (async) ----------
import asyncio

async def export_photo_index_excel():
    global row_ptr

    limit = TEST_LIMIT if TEST_MODE else None
    processed = 0
    skipped = 0

    print(f"🧪 TEST_MODE={TEST_MODE} | limit={limit if limit is not None else 'FULL'}")

    for file_id, figma_pages, page_names in file_page_sets:
        for page_name in page_names:
            page = find_page(figma_pages, page_name)
            if not page:
                print(f"⚠️ Page not found: {page_name}")
                continue

            assign_parents(page)

            module_number = parse_module_number(page_name)
            print(f"\n📄 {page_name} | module={module_number} | file={file_id}")

            for node in walk(page):
                image_ref = is_image_rectangle(node)
                if not image_ref:
                    continue

                node_id = node["id"]
                photo_layer_name = node.get("name")  # ✅ photo layer name

                # Your notebook’s fetch_image_url_single signature is: (file_id, node_id, canvas_height)
                canvas_height = int((node.get("absoluteBoundingBox") or {}).get("height") or 0) or 600
                url = fetch_image_url_single(file_id, node_id, canvas_height)
                if not url:
                    skipped += 1
                    continue

                img_bytes = await fetch_image_content(url)
                if not img_bytes:
                    skipped += 1
                    continue

                # hierarchy info
                lesson_section = find_lesson_section(node)
                chapter_section = find_chapter_section_from_lesson(lesson_section)

                lesson_section_name = lesson_section.get("name") if lesson_section else None
                chapter_section_name = chapter_section.get("name") if chapter_section else None

                chapter_number = parse_chapter_number(chapter_section_name or "")
                lesson_number  = parse_lesson_number(lesson_section_name or "")

                frame = find_nearest_ancestor(node, ("FRAME",))
                frame_name = frame.get("name") if frame else None

                # thumbnail
                thumb_bio, (tw, th) = make_jpeg_thumbnail_bytes(
                    img_bytes, max_px=THUMB_MAX_PX, quality=THUMB_QUALITY
                )

                # Write metadata (thumbnail is column A but embedded, so we write other fields starting col B)
                ws.cell(row_ptr, 2, file_id)
                ws.cell(row_ptr, 3, module_number)
                ws.cell(row_ptr, 4, chapter_number)
                ws.cell(row_ptr, 5, lesson_number)
                ws.cell(row_ptr, 6, page_name)
                ws.cell(row_ptr, 7, chapter_section_name)
                ws.cell(row_ptr, 8, lesson_section_name)
                ws.cell(row_ptr, 9, frame_name)
                ws.cell(row_ptr, 10, photo_layer_name)
                ws.cell(row_ptr, 11, node_id)
                ws.cell(row_ptr, 12, image_ref)
                ws.cell(row_ptr, 13, url)

                # Embed thumbnail in column A
                xl_img = XLImage(thumb_bio)
                xl_img.width, xl_img.height = tw, th
                xl_img.anchor = f"A{row_ptr}"
                ws.add_image(xl_img)
                ws.row_dimensions[row_ptr].height = px_to_excel_row_height(th + 6)

                print(f"  ✅ [{processed+1}{f'/{limit}' if limit else ''}] "
                      f"M{module_number} C{chapter_number} L{lesson_number} | "
                      f"layer='{photo_layer_name or ''}' | frame='{frame_name or ''}'")

                row_ptr += 1
                processed += 1

                if limit is not None and processed >= limit:
                    os.makedirs(os.path.dirname(EXCEL_OUT_PATH), exist_ok=True)
                    wb.save(EXCEL_OUT_PATH)
                    print(f"\n✅ Stopped after {processed} images (test). Saved: {EXCEL_OUT_PATH}")
                    print(f"   Skipped: {skipped}")
                    return

    os.makedirs(os.path.dirname(EXCEL_OUT_PATH), exist_ok=True)
    wb.save(EXCEL_OUT_PATH)
    print(f"\n✅ Finished FULL run. Processed: {processed} | Skipped: {skipped}")
    print(f"📎 Saved: {EXCEL_OUT_PATH}")

await export_photo_index_excel()
```

```python
# ✅ STANDALONE: CLEAN INDEX (chart + infographic) — EXPORT WHOLE TEMPLATE FRAME VISUALLY (ROBUST)
# Supports 3 figma files.
#
# Row identity (unchanged):
#   - chart: each IMAGE-filled RECTANGLE inside 'chart' template (id = rectangle name)
#   - infographic: each direct child FRAME inside 'infographic' template (id = inner frame name)
#
# Thumbnail export (robust):
#   - export the FULL template root frame (Chart / infographic)
#   - if PNG/JPG render returns null, fallback to PDF export and rasterize locally
#
# Output columns: thumbnail | id | lesson (M.C.L) | type

import os, re, json, random, time
from io import BytesIO
from PIL import Image
from openpyxl import Workbook
from openpyxl.drawing.image import Image as XLImage
import asyncio

import requests

# =========================
# 🔧 SWITCHES
# =========================
TEST_MODE          = True
TEST_SAMPLE_SIZE   = 8
RANDOM_SEED        = 7
EXCEL_OUT_PATH     = "../03_Outputs/index_chart_infographic_clean.xlsx"

THUMB_MAX_PX       = (360, 360)
THUMB_QUALITY      = 78

WRITE_SKIPPED_SHEET = True
MAX_LOG_SKIPS        = 50

# PDF rasterization quality (only used on fallback)
PDF_ZOOM = 2.0  # 2.0 ~ good thumbnails; increase if you need sharper
# =========================


# ---------- Load cached Figma JSONs (3 files) ----------
with open(figma_path_1, "r") as f:
    figma_data_1 = json.load(f)
with open(figma_path_2, "r") as f:
    figma_data_2 = json.load(f)
with open(figma_path_3, "r") as f:
    figma_data_3 = json.load(f)

file_page_sets = [
    (FILE_ID_1, figma_data_1["children"], pages_1),
    (FILE_ID_2, figma_data_2["children"], pages_2),
    (FILE_ID_3, figma_data_3["children"], pages_3),
]


# ---------- Utilities ----------
def px_to_excel_row_height(px): return px * 0.75
def px_to_excel_col_width(px): return px / 7.0
def norm(s): return (s or "").strip().lower()

def safe_str(x, maxlen=240):
    s = "" if x is None else str(x)
    return s if len(s) <= maxlen else s[:maxlen] + "…"

def make_thumb(image_bytes):
    with Image.open(BytesIO(image_bytes)) as im:
        im = im.convert("RGB")
        im.thumbnail(THUMB_MAX_PX)
        bio = BytesIO()
        im.save(bio, format="JPEG", quality=THUMB_QUALITY, optimize=True, progressive=True)
        bio.seek(0)
        return bio, im.size

def get_abs_height(node, fallback=900):
    bb = node.get("absoluteBoundingBox") or {}
    h = bb.get("height")
    try:
        h = int(h)
    except Exception:
        h = None
    return h if h and h > 0 else fallback


# ---------- Tree helpers ----------
def assign_parents(node, parent=None):
    node["_parent"] = parent
    for c in node.get("children", []) or []:
        assign_parents(c, node)

def walk(node):
    yield node
    for ch in node.get("children", []) or []:
        yield from walk(ch)

def find_nearest_ancestor(node, types):
    cur = node.get("_parent")
    while cur:
        if cur.get("type") in types:
            return cur
        cur = cur.get("_parent")
    return None

def find_next_ancestor_section(start_node):
    cur = start_node.get("_parent") if start_node else None
    while cur:
        if cur.get("type") == "SECTION":
            return cur
        cur = cur.get("_parent")
    return None

def ancestry_chain(node, max_depth=12):
    chain = []
    cur = node
    for _ in range(max_depth):
        if not cur: break
        chain.append(f"{cur.get('type')}:{cur.get('name')}")
        cur = cur.get("_parent")
    return " <- ".join(chain)


# ---------- Parsing ----------
def parse_module_number(name):
    if not name: return None
    m = re.search(r"\bModule\s*0*(\d+)\b", name, re.IGNORECASE)
    if m: return m.group(1)
    m = re.search(r"\bM\s*0*(\d+)\b", name, re.IGNORECASE)
    if m: return m.group(1)
    return None

def parse_chapter_number(name):
    if not name: return None
    m = re.search(r"\bChapter\s*0*(\d+)\b", name, re.IGNORECASE)
    if m: return m.group(1)
    m = re.search(r"\bC\s*0*(\d+)\b", name, re.IGNORECASE)
    if m: return m.group(1)
    return None

def parse_lesson_number(name):
    if not name: return None
    m = re.search(r"\bLesson\s*0*(\d+)\b", name, re.IGNORECASE)
    if m: return m.group(1)
    m = re.search(r"\bL\s*0*(\d+)\b", name, re.IGNORECASE)
    if m: return m.group(1)
    return None

def get_lesson_triplet(node):
    lesson_section = find_nearest_ancestor(node, ("SECTION",))
    while lesson_section and not parse_lesson_number(lesson_section.get("name","")):
        lesson_section = find_next_ancestor_section(lesson_section)

    chapter_section = find_next_ancestor_section(lesson_section) if lesson_section else None
    while chapter_section and not parse_chapter_number(chapter_section.get("name","")):
        chapter_section = find_next_ancestor_section(chapter_section)

    page = find_nearest_ancestor(node, ("CANVAS",))
    module_number = parse_module_number(page.get("name") if page else "")

    lesson_number  = parse_lesson_number(lesson_section.get("name") if lesson_section else "")
    chapter_number = parse_chapter_number(chapter_section.get("name") if chapter_section else "")

    if module_number and chapter_number and lesson_number:
        return f"{module_number}.{chapter_number}.{lesson_number}"
    return None


# ---------- Template detection ----------
def is_image_rectangle(node):
    if node.get("type") != "RECTANGLE":
        return False
    for f in node.get("fills") or []:
        if f.get("type") == "IMAGE" and f.get("imageRef"):
            return True
    return False

def is_template_root_frame(node):
    # export roots are frames named chart/infographic (case-insensitive)
    if node.get("type") != "FRAME":
        return None
    n = norm(node.get("name"))
    return n if n in {"chart", "infographic"} else None


# ---------- Figma export helpers ----------
def figma_images_api_url(file_id: str, node_id: str, *, fmt: str, scale: int = 2, use_absolute_bounds: bool = True):
    api = f"https://api.figma.com/v1/images/{file_id}"
    headers = {"X-Figma-Token": FIGMA_TOKEN}
    params = {"ids": node_id, "format": fmt, "scale": scale}
    if use_absolute_bounds:
        params["use_absolute_bounds"] = "true"

    t0 = time.time()
    r = requests.get(api, headers=headers, params=params, timeout=60)
    ms = (time.time() - t0) * 1000

    if r.status_code != 200:
        return None, f"HTTP {r.status_code} (ms={ms:.1f}) body={r.text[:250]}"
    data = r.json()
    images = data.get("images") or {}
    url = images.get(node_id)
    if not url:
        # This is your case: node id exists but URL is null
        return None, f"render_null (ms={ms:.1f}) err={data.get('err')} images_val={images.get(node_id)}"
    return url, f"ok (ms={ms:.1f})"

async def fetch_bytes(url: str) -> bytes | None:
    try:
        return await fetch_image_content(url)
    except Exception:
        return None

def rasterize_pdf_first_page_to_png_bytes(pdf_bytes: bytes, zoom: float = 2.0) -> bytes:
    """
    Rasterize PDF first page to PNG bytes using PyMuPDF (fitz).
    """
    try:
        import fitz  # PyMuPDF
    except Exception as e:
        raise RuntimeError(
            "PDF fallback requires PyMuPDF (pip install pymupdf) to rasterize."
        ) from e

    doc = fitz.open(stream=pdf_bytes, filetype="pdf")
    page = doc.load_page(0)
    mat = fitz.Matrix(zoom, zoom)
    pix = page.get_pixmap(matrix=mat, alpha=False)
    return pix.tobytes("png")


async def export_node_visual_bytes(file_id: str, node_id: str) -> tuple[bytes | None, str]:
    """
    Try to obtain a visual image of the node.
    Strategy: PNG -> JPG -> PDF (then rasterize).
    Returns: (bytes, debug_status)
    """
    # 1) PNG
    url, dbg = figma_images_api_url(file_id, node_id, fmt="png", scale=2, use_absolute_bounds=True)
    if url:
        b = await fetch_bytes(url)
        if b:
            return b, f"png:{dbg}"

    # 2) JPG (sometimes succeeds when png fails)
    url, dbg2 = figma_images_api_url(file_id, node_id, fmt="jpg", scale=2, use_absolute_bounds=True)
    if url:
        b = await fetch_bytes(url)
        if b:
            return b, f"jpg:{dbg2}"

    # 3) PDF export (often succeeds when raster fails), then rasterize locally
    url, dbg3 = figma_images_api_url(file_id, node_id, fmt="pdf", scale=1, use_absolute_bounds=True)
    if url:
        pdf_bytes = await fetch_bytes(url)
        if pdf_bytes:
            try:
                png_bytes = rasterize_pdf_first_page_to_png_bytes(pdf_bytes, zoom=PDF_ZOOM)
                return png_bytes, f"pdf->png:{dbg3}"
            except Exception as e:
                return None, f"pdf_rasterize_failed:{dbg3} err={repr(e)}"

    return None, f"failed png({dbg}) jpg({dbg2}) pdf({dbg3})"


# ---------- Collect candidates ----------
def collect_candidates():
    candidates = []

    for file_id, figma_pages, page_names in file_page_sets:
        for page_name in page_names:
            page = next((p for p in figma_pages if p.get("name") == page_name), None)
            if not page:
                print(f"⚠️ Page not found in cached JSON: {page_name}")
                continue

            assign_parents(page)

            for node in walk(page):
                root_type = is_template_root_frame(node)
                if not root_type:
                    continue

                template_root = node

                if root_type == "chart":
                    for sub in walk(template_root):
                        if is_image_rectangle(sub):
                            candidates.append({
                                "type": "chart",
                                "file_id": file_id,
                                "node": sub,                 # row identity (rectangle)
                                "export_root": template_root # export the whole chart frame visually
                            })

                elif root_type == "infographic":
                    for sub in template_root.get("children", []) or []:
                        if sub.get("type") == "FRAME":
                            candidates.append({
                                "type": "infographic",
                                "file_id": file_id,
                                "node": sub,                 # row identity (inner frame)
                                "export_root": template_root # export the whole infographic frame visually
                            })

    return candidates


# ---------- Workbook ----------
wb = Workbook()
ws = wb.active
ws.title = "Index"
ws.append(["thumbnail", "id", "lesson", "type"])
ws.column_dimensions["A"].width = px_to_excel_col_width(THUMB_MAX_PX[0] + 14)

ws_skipped = None
if WRITE_SKIPPED_SHEET:
    ws_skipped = wb.create_sheet("Skipped")
    ws_skipped.append(["type", "id", "lesson", "reason", "file_id", "export_root_id", "detail", "ancestry"])


# ---------- Export ----------
async def export_index():
    candidates = collect_candidates()
    total = len(candidates)

    if RANDOM_SEED is not None:
        random.seed(RANDOM_SEED)

    if TEST_MODE:
        sample_size = min(TEST_SAMPLE_SIZE, total)
        candidates = random.sample(candidates, sample_size)
        print(f"🧪 TEST MODE — sampling {sample_size} of {total}")
    else:
        print(f"🚀 FULL RUN — processing {total}")

    processed = 0
    skipped = 0
    skip_reasons = {}
    row_ptr = 2

    def note_skip(item, reason, detail=""):
        nonlocal skipped
        skipped += 1
        skip_reasons[reason] = skip_reasons.get(reason, 0) + 1
        if skipped <= MAX_LOG_SKIPS:
            print(f"  ⚠️ SKIP [{skipped}] {item['type']} | id='{item['node'].get('name')}' | lesson='{item.get('lesson')}' | {reason} {safe_str(detail)}")
        if ws_skipped is not None:
            ws_skipped.append([
                item["type"],
                item["node"].get("name"),
                item.get("lesson"),
                reason,
                item["file_id"],
                item["export_root"].get("id") if item.get("export_root") else None,
                safe_str(detail, 800),
                ancestry_chain(item["node"], 18),
            ])

    for item in candidates:
        node = item["node"]
        export_root = item["export_root"]
        file_id = item["file_id"]

        asset_id = node.get("name")  # row id rule unchanged
        lesson_triplet = get_lesson_triplet(node)
        item["lesson"] = lesson_triplet

        if not lesson_triplet:
            note_skip(item, "missing_lesson_triplet")
            continue
        if not asset_id:
            note_skip(item, "missing_id_name")
            continue

        export_node_id = export_root.get("id")
        if not export_node_id:
            note_skip(item, "missing_export_root_id")
            continue

        # ✅ export whole frame visually with robust fallback
        img_bytes, dbg = await export_node_visual_bytes(file_id, export_node_id)
        if not img_bytes:
            note_skip(item, "render_failed_all_formats", dbg)
            continue

        try:
            thumb_bio, (tw, th) = make_thumb(img_bytes)
        except Exception as e:
            note_skip(item, "thumb_exception", repr(e))
            continue

        ws.cell(row_ptr, 2, asset_id)
        ws.cell(row_ptr, 3, lesson_triplet)
        ws.cell(row_ptr, 4, item["type"])

        xl_img = XLImage(thumb_bio)
        xl_img.width, xl_img.height = tw, th
        xl_img.anchor = f"A{row_ptr}"
        ws.add_image(xl_img)
        ws.row_dimensions[row_ptr].height = px_to_excel_row_height(th + 6)

        processed += 1
        print(f"  ✅ {processed}: {item['type']} | {lesson_triplet} | id='{asset_id}' (thumb={dbg})")
        row_ptr += 1

    os.makedirs(os.path.dirname(EXCEL_OUT_PATH), exist_ok=True)
    wb.save(EXCEL_OUT_PATH)

    print(f"\n✅ Done. Saved: {EXCEL_OUT_PATH}")
    print(f"📊 Processed: {processed} | Skipped: {skipped}")
    if skipped:
        print("📌 Skip reasons:")
        for k, v in sorted(skip_reasons.items(), key=lambda x: (-x[1], x[0])):
            print(f"  - {k}: {v}")

await export_index()
```
